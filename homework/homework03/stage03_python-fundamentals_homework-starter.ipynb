## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit
```

```python
import numpy as np

arr = np.arange(1, 1000001)

def loop_double(x):
    result = []
    for value in x:
        result.append(value * 2)
    return result

def vectorized_double(x):
    return x * 2

print("Loop:")
%timeit loop_double(arr)

print("Vectorized:")
%timeit vectorized_double(arr)
```

## 4. Summary Statistics and Groupby

```python
# TODO: Perform groupby and save output to data/processed/
```

```python
from pathlib import Path

# Summary statistics for numeric columns
summary = df.describe()

print("Summary Statistics:")
display(summary)

# Check available columns
print("\nColumns in dataset:")
print(df.columns.tolist())

# Groupby aggregation by category
if "category" in df.columns:
    grouped = df.groupby("category").mean(numeric_only=True)

    print("\nGroupby Results:")
    display(grouped)
else:
    print("\nNo 'category' column found. Please check the actual column name above.")

# Create processed folder
processed_dir = Path("data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

# Save summary statistics
summary_path = processed_dir / "summary.csv"
summary.to_csv(summary_path)

print("\nSaved summary to:")
print(summary_path.resolve())
```
